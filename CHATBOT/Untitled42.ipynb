```python
from urllib.parse import urlparse, parse_qs
```

```python
#basically we will bring the document into our vectorstore
def get_transcript_from_url(video_url):
    try:
        # 1. Extract Video ID (Handles standard and short URLs)
        parsed_url = urlparse(video_url)
        if "youtu.be" in video_url:
            video_id = parsed_url.path[1:]
        else:
            video_id = parse_qs(parsed_url.query).get("v")[0]
           
    #we will try to get the youtube transcript form the youtube id by using the youtube video transcript

        fetched_transcript=YouTubeTranscriptApi().fetch(video_id,languages=["en"])
  # FetchedTranscript(snippets=[FetchedTranscriptSnippet(text='Enhancement of construction and', start=6.799, duration=4.96), FetchedTranscriptSnippet(text='infrastructure equipments, high-speed', start=8.96, duration=6.24), FetchedTranscriptSnippet(text='rail corridors as growth connectors,', start=11.759, duration=6.321), FetchedTranscriptSnippet(text='individual persons resident outside', start=15.2, duration=5.68), FetchedTranscriptSnippet(text='India will be permitted to invest in', start=18.08, duration=5.279), FetchedTranscriptSnippet(text='equity instruments, medical tourism': 2.0}
   #the above is the one which is expected ny the above
        transcript_list=fetched_transcript.to_raw_data()
   #the above step return a list so we will flattent that into a plain text
        transcript=" ".join([d["text"] for d in transcript_list])
        return transcript
    except TranscriptsDisabled:
        print("Transcripts are disabled for this")
```

```python
splitter=RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
    )
```

```python
chunks=splitter.split_text(transcript)
```

```python
print(len(chunks))
```

```python
chunks[0]
```

```python
model=GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")
```

```python
vector_store=Chroma(
    embedding_function=model,
    persist_directory="./youtube_transcripts",
    collection_name="youtube_transcripts"
)
```

```python
vector_store.add_texts(chunks)
```

```python
vector_store.get(include=["embeddings","metadatas","documents"])
```

```python
vector_store.get(ids=['e9049194-3d72-4313-bcb5-f5938b18c3dd'])
```

**Retriever**

```python
retriever=vector_store.as_retriever(search_type="similarity",search_kwargs={"k":4})
```

```python
retriever
```

```python
result=retriever.invoke("What this video is all about ?")
```

```python
print(result)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt=PromptTemplate(
    template="""
    you are a helpful AI assistant ,
    Answer Only From the provided transcript context.If the context is just insufficient for the asked question just say you don't know .\n {context} \n question: {question}
    """,
    input_variables=["context","question"])

```

**Making the MODEL**

```python
model=ChatGoogleGenerativeAI(model="gemini-2.5-flash",temperature=0.2)
```

```python
question="Is This video is discussing about the Budget of 2026 of India .if yes then what are teh key points?"
```

```python
retreived_docs=retriever.invoke(question)
```

```python
context_text=" ".join(doc.page_content for doc in retreived_docs)
```

```python
final_prompt=prompt.invoke({
    "context":context_text,
    "question":question
})
```

**Generating the response **

```python
result=model.invoke(final_prompt)
```

```python
print(result.content)
```

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough,RunnableLambda
```

```python
def format_text(text):
  context_text=" ".join(doc.page_content for doc in text)
  return context_text
```

```python
parallel_chain=RunnableParallel({
    "context":retriever|RunnableLambda(format_text),
    "question":RunnablePassthrough()
})
```

```python
parallel_chain.invoke("Is This video is discussing about the Budget of 2026 of India .if yes then what are teh key points?")
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser=StrOutputParser()
```

```python
sequence_chain=prompt|model|parser
```

```python
final_chain=parallel_chain|sequence_chain
```

```python


sequence_chain=prompt|model|parser
final_chain=parallel_chain|sequence_chain

ans=final_chain.invoke({
    "Is this video is discussing about the Budget of 2026 of India .if yes then what are teh key points?"
})
```

```python
print(ans)
```
